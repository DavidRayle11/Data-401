## 4) Develop Analytical Questions
### 3 Analytical Questions:
1. How are audio characteristics energy, valence, loudness, and acousticness related to a song's popularity?
    - **Variables:** popularity, energy, valence, loudness, acousticness.
    - **Why it matters:** This question investigates whether certain musical characteristics are associated with more popular songs.
    - **Analysis:** Correlations and scatterplots can be used to examine the strength and direction of relationships between popularity and audio characteristics.

2. How do musical characteristics and popularity differ across Spotify genres?
    - **Variables:** track_genre, popularity, danceability, energy, valence, acousticness.
    - **Why it matters:** Different genres have distinct musical characteristics and audiences. 
    Comparing genres can reveal whether certain tend too be more popular and how their audio characteristics differ.
    - **Analysis:** Grouped summary statistics, bar charts, and boxplots can be used to compare genres.

3. Do explicit and non-explicit songs differ in popularity, danceability, and energy?
    - **Variables:** explicit, popularity, danceability, energy.
    - **Why it matters:** Explicit content provides a categorical characteristic that can be compared with both popularity and musical characteristics.
    - **Analysis:** Grouped summary statistics and boxplots can be used to compare the distribuitons of explicit and non-explicit tracks.


### Visualization 1)

```python
plt.figure(figsize=(10,6))
sns.histplot(data=spotify_unique, x="popularity", bins=20)

plt.title("Distribution of Spotify Track Popularity")
plt.xlabel("Popularity Score")
plt.ylabel("Number of Tracks")

plt.show()
```

```python
spotify_unique["popularity"].describe()
```

```python
spotify_unique["popularity"].median()
```

```python
## Interpretation
# The distribution shows that Spotify track popularity varies considerably, with a median popularity of 33. Most tracks are concentrated in the lower-to-middlerange of popularity score, while
# relatively few tracks have popularity scores above 80. There is a noticeable concentration of tracks near a popularity score of 0, 
# this may represents need to see whether these represent genuinely unpopular tracks or another characteristic of Spotify's popularity measure.
# This is a right-skewed distribution
```

### Visualization 2)

```python
audio_features = [
    "popularity",
    "danceability",
    "energy",
    "valence",
    "loudness",
    "acousticness",
    "speechiness",
    "instrumentalness",
    "liveness",
    "tempo"
]

spotify_unique[audio_features].corr()["popularity"].sort_values(ascending=False)
```

```python
pop_corr = (
    spotify_unique[audio_features]
    .corr()["popularity"]
    .drop("popularity")
    .sort_values(ascending=False)
)

print(pop_corr)
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=spotify_unique,
    x="instrumentalness",
    y="popularity",
    alpha=0.3
)

plt.title("Spotify Track Popularity vs. Instrumentalness")
plt.xlabel("Instrumentalness")
plt.ylabel("Popularity Score")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.regplot(
    data=spotify_unique,
    x="instrumentalness",
    y="popularity",
    scatter_kws={"alpha": 0.15},
    line_kws={"color": "red"}
)

plt.title("Relationship Between Instrumentalness and Spotify Popularity")
plt.xlabel("Instrumentalness")
plt.ylabel("Popularity Score")

plt.show()
```

Interpretation:
There is an overall downward slope to this regression line, however I do not think that this downward slope means that there is anything in particular happening within the regression line that has been given. The instrumentalness alone does not explain popularity. the correlation was -.127 indicating a weak negative relationship

### Visualization 3 -- Genre and Popularity

```python
avg_genre_popularity = (
    spotify.groupby("track_genre")["popularity"]
    .mean()
    .sort_values(ascending=False)
)

avg_genre_popularity.head(15)
```

```python
## let us see the highest and the lost of the average genre popularity
avg_genre_popularity.tail(15)
```

```python
top10genres = avg_genre_popularity.head(10)

plt.figure(figsize=(10,6))

sns.barplot(
    x=top10genres.values,
    y=top10genres.index
)

plt.title("Top 10 Spotify Genres by Average Track Popularity")
plt.xlabel("Average Popularity Score")
plt.ylabel("Track Genre")

plt.show()
```

```python
avg_genre_popularity.head(10).round(2)
```

Interpretation
the broader genre ranks 10th, 1st is pop-film, followed by k-pop and chill, this suggests that there are diferences that genre may provide more useful for contextual understanding of popularity than the individual audio characteristics. 

### Visualization 4

```python
explicit_stats = (
    spotify_unique.groupby("explicit")["popularity"]
    .agg(["count", "mean", "median"])
    .round(2)
)

explicit_stats
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(
    data=spotify_unique, x="explicit", y="popularity"
)

plt.title("Popularity of Explicit vs Non-Explicit Spotify Tracks")
plt.xlabel("Explicit Content")
plt.ylabel("Popularity Score")

plt.show()
```

Interpretation: 
the explicit tracks show that a slightly higher median popularity than non-explicit tracks. 

while the distributions overlap substantiallythis also indicates that they both have a wide range of popularity scoress. 

This suggests an association between explicit status and popularity in this dataset, but the content alone does not seem to be a strong enough variable to distinguish the track popularity.
